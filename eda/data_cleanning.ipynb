```python
import pandas as pd
import os
```

```python
path_client_input = 'data/clean_fruki_data/dim_cliente.csv'
path_client_output = 'data/clean_fruki_data/dim_cliente_clean.csv'

data_client = pd.read_csv(path_client_input)

useful_columns = [
    'id_cliente', 'id_hierarquia', 'ID_TABELA_PRECO', 
    'ESTADO', 'CIDADE', 'bairro', 'segmento', 'SUBSEGMENTO', 
    'situacao_cliente', 'SITUACAO_CREDITO',
    'ENDERECO', 'CEP', 'LATITUDE', 'LONGITUDE'
]

data_client = data_client[useful_columns]
data_client.to_csv(path_client_output, index=False)

print(f"{data_client.shape[1]} colunas e {data_client.shape[0]} linhas.")
```

```python
path_hierarquia_input = 'data/clean_fruki_data/dim_hierarquia_comercial.csv'
path_hierarquia_output = 'data/clean_fruki_data/dim_hierarquia_comercial_clean.csv'

data_hierarquia = pd.read_csv(path_hierarquia_input)

useful_columns_hierarquia = [
    'ID_HIERARQUIA', 'COD_REPRESENTANTE', 'REPRESENTANTE'
]

data_hierarquia = data_hierarquia[useful_columns_hierarquia]
data_hierarquia.to_csv(path_hierarquia_output, index=False)

print(f"{data_hierarquia.shape[1]} colunas e {data_hierarquia.shape[0]} linhas.")
```

```python
path_item_input = 'data/clean_fruki_data/dim_item.csv'
path_item_output = 'data/clean_fruki_data/dim_item_clean.csv'

data_item = pd.read_csv(path_item_input)

useful_columns_item = [
    'ID_ITEM', 'COD_DESC_FAMILIA', 'COD_DESC_LINHA_PROD', 
    'RESERVA', 'OBSOLETO', 'SABOR', 'EMBALAGEM'
]

data_item = data_item[useful_columns_item]
data_item.to_csv(path_item_output, index=False)

print(f"{data_item.shape[1]} colunas e {data_item.shape[0]} linhas.")
```

```python
path_preco_input = 'data/clean_fruki_data/dim_tabela_preco.csv'
path_preco_output = 'data/clean_fruki_data/dim_tabela_preco_clean.csv'

data_preco = pd.read_csv(path_preco_input)

useful_columns_preco = [
    'TABELA_PRECO', 'ID_ITEM', 'PRECO_VENDA', 'PRECO_MINIMO', 'TABELA_ATIVA'
]

data_preco = data_preco[useful_columns_preco]
data_preco.to_csv(path_preco_output, index=False)

print(f"{data_preco.shape[1]} colunas e {data_preco.shape[0]} linhas.")
```

```python
path_client_input = 'data/clean_fruki_data/dim_cliente_clean.csv'
path_client_output = 'data/clean_fruki_data/dim_cliente_clean.csv'

data_client = pd.read_csv(path_client_input)

data_client = data_client[
    (data_client['situacao_cliente'] == 'Ativo') & 
    (data_client['SITUACAO_CREDITO'] != 'SUSPENSO')
]

data_client = data_client.drop(columns=['situacao_cliente', 'SITUACAO_CREDITO'])
data_client.to_csv(path_client_output, index=False)

print(f"restantes: {len(data_client)}")
```

```python
path_item_input = 'data/clean_fruki_data/dim_item_clean.csv'
path_item_output = 'data/clean_fruki_data/dim_item_clean.csv'

data_item = pd.read_csv(path_item_input)

data_item = data_item[data_item['OBSOLETO'] == 'Obs_1512816538']
data_item = data_item.drop(columns=['OBSOLETO'])
data_item.to_csv(path_item_output, index=False)

print(f"Linhas restantes na dim_item: {len(data_item)}")
```

```python
path_preco_input = 'data/clean_fruki_data/dim_tabela_preco_clean.csv'
path_preco_output = 'data/clean_fruki_data/dim_tabela_preco_clean.csv'

df_preco = pd.read_csv(path_preco_input)

df_preco = df_preco[df_preco['TABELA_ATIVA'] == 'Ativo']
df_preco = df_preco.drop(columns=['TABELA_ATIVA'])
df_preco.to_csv(path_preco_output, index=False)

print(f"restantes na dim_tabela_preco: {len(df_preco)}")
```

```python

paths = {
    'Clientes': 'data/clean_fruki_data/dim_cliente_clean.csv',
    'Itens': 'data/clean_fruki_data/dim_item_clean.csv',
    'Preços': 'data/clean_fruki_data/dim_tabela_preco_clean.csv',
    'Hierarquia': 'data/clean_fruki_data/dim_hierarquia_comercial_clean.csv' 
}

for name, path in paths.items():
    if os.path.exists(path):
        df = pd.read_csv(path)
        print(f"--- {name} ---")
        print(f"Total de Linhas: {df.shape[0]}")
        print(f"Total de Colunas: {df.shape[1]}")
        print(f"Colunas: {', '.join(df.columns.tolist())}\n")
    else:
        print(f"--- {name} ---")
        print(f"Aviso: Arquivo '{path}' não encontrado. Verifique se o caminho está correto.\n")
```
