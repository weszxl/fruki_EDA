```python
import pandas as pd
import numpy as np
from datetime import datetime, timedelta
import random

df = pd.read_csv('data/cliente.csv')

num_clientes = len(df)

def generate_cnpj():
    return f"{random.randint(10, 99)}.{random.randint(100, 999)}.{random.randint(100, 999)}/0001-{random.randint(10, 99)}"

df['cnpj'] = [generate_cnpj() for _ in range(num_clientes)]

nomes_base = ['Silva', 'Central', 'Oliveira', 'Santos', 'Avenida', 'Sul', 'Gaucho', 'Ponto Certo', 'Esquina', 'Amizade']
df['nome'] = df['subsegmento'].astype(str) + ' ' + np.random.choice(nomes_base, num_clientes)

def generate_celular():
    return f"519{random.randint(8000,9999)}{random.randint(1000,9999)}"

df['numero_celular'] = [generate_celular() for _ in range(num_clientes)]

now = datetime.now()
def random_date():
    days_ago = random.randint(1, 60)
    hours_ago = random.randint(0, 23)
    minutes_ago = random.randint(0, 59)
    return (now - timedelta(days=days_ago, hours=hours_ago, minutes=minutes_ago)).strftime('%Y-%m-%d %H:%M:%S')

df['ultima_compra'] = [random_date() for _ in range(num_clientes)]

cidades_unicas = df[''].unique()
df_cidade = pd.DataFrame({
    'id_cidade': range(1, len(cidades_unicas) + 1),
    'nome': cidades_unicas,
    'uf': 'RS' 
})

df = df.merge(df_cidade, left_on='cidade_original', right_on='nome', how='left')

colunas_finais_cliente = [
    'id_cliente', 'cnpj', 'nome_x', 'segmento', 'subsegmento', 
    'numero_celular', 'cep', 'ultima_compra', 'id_hierarquia', 'id_cidade'
]

df = df.rename(columns={'nome_x': 'nome'})

df_final = df[['id_cliente', 'cnpj', 'nome', 'segmento', 'subsegmento', 'numero_celular', 'cep', 'ultima_compra', 'id_hierarquia', 'id_cidade']]

df_final.to_csv('cliente_pronto.csv', index=False)
df_cidade.to_csv('cidade_pronta.csv', index=False)
```
