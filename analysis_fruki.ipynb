```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

#file_path = 'data/fruki_data/'
#data = pd.read_csv(file_path)
pd.set_option('display.max_columns', None)

data = [
    'data/fruki_data/dim_cliente.csv', 
    'data/fruki_data/dim_hierarquia_comercial.csv', 
    'data/fruki_data/dim_item.csv', 
    'data/fruki_data/dim_tabela_preco.csv'
]
```

```python

for fruki_columns in data:
    try:
        df = pd.read_csv(fruki_columns, nrows=0, sep=None, engine='python')
        print(f"--- {fruki_columns} ---")
        print(list(df.columns))
        print("\n")
    except Exception as e:
        print(f"{fruki_columns}: {e}")
```

```python
for fruki_items in data:
    try:
        df = pd.read_csv(fruki_items, nrows=1, sep=None, engine='python')
        print(f"--- {fruki_items} ---")
        
        if not df.empty:
            for coluna, valor in df.iloc[0].items():
                print(f"{coluna}: {valor}")
        else:
            print("T-T")
            
        print("\n" + "="*40 + "\n")
        
    except Exception as e:
        print(f"Erro ao ler {fruki_items}: {e}")
```

- Validar se há risco de duplicação. O código testa se as tabelas de dimensões guardam versões antigas dos cadastros (se um produto mudou de preço, ele gera uma linha nova ou substitui a velha?). 

- Validar o tamanho do público, medidndo a base de clientes. A ideia era descobrir qual o volume real de contas disponíveis para consumir as recomendações.

```python
dim_cliente = pd.read_csv(data[0], sep=None, engine='python')
dim_hierarquia = pd.read_csv(data[1], sep=None, engine='python')
dim_item = pd.read_csv(data[2], sep=None, engine='python')
dim_preco = pd.read_csv(data[3], sep=None, engine='python')

print("--- Base ---")
print(f"Clientes: {len(dim_cliente)}")
print(f"Hierarquia: {len(dim_hierarquia)}")
print(f"Itens: {len(dim_item)}")
print(f"Preços: {len(dim_preco)}\n")

current_date = '2199-12-31T00:00:00.000Z'
dim_hierarquia_ativa = dim_hierarquia[dim_hierarquia['DATE_TO'] == current_date]
dim_item_ativo = dim_item[dim_item['DATE_TO'] == current_date]
dim_preco_ativo = dim_preco[dim_preco['DATE_TO'] == current_date]

clientes = dim_cliente['situacao_cliente'].value_counts()

print("--- Restante (apenas ativos) ---")
print(f"Hierarquia: {len(dim_hierarquia_ativa)}")
print(f"Itens: {len(dim_item_ativo)}")
print(f"Preços: {len(dim_preco_ativo)}\n")

print("--- Situação dos Clientes ---")
print(clientes)
```
